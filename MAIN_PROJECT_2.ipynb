Task 128: Create an Excel file with multiple sheets:

Sheet 1: Full cleaned data

Sheet 4: Top 50 customers

```python
onre['Sales'] = onre['Quantity'] * onre['UnitPrice']
```

```python
sheet1 = onre
```

```python
sheet2 = onre.groupby('Country').agg({
    'Sales': 'sum',
    'Quantity': 'sum'
}).reset_index()
```

```python
onre['InvoiceDate'] = pd.to_datetime(onre['InvoiceDate'])
sheet3 = onre.groupby(onre['InvoiceDate'].dt.to_period('M'))['Sales'].sum().reset_index()
sheet3['InvoiceDate'] = sheet3['InvoiceDate'].astype(str)
```

```python
sheet4 = (
    onre.groupby('CustomerID')['Sales']
    .sum()
    .reset_index()
    .sort_values(by='Sales', ascending=False)
    .head(50)
)
```
